```python
import pandas as pd
import requests
from io import StringIO

def get_ipos_by_year(year: int) -> pd.DataFrame:
    """
    Fetch IPO data for the given year from stockanalysis.com.
    """
    url = f"https://stockanalysis.com/ipos/{year}/"
    headers = {
        'User-Agent': (
            'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
            'AppleWebKit/537.36 (KHTML, like Gecko) '
            'Chrome/58.0.3029.110 Safari/537.3'
        )
    }

    try:
        response = requests.get(url, headers=headers, timeout=10)
        response.raise_for_status()

        # Wrap HTML text in StringIO to avoid deprecation warning
        # "Passing literal html to 'read_html' is deprecated and will be removed in a future version. To read from a literal string, wrap it in a 'StringIO' object."
        html_io = StringIO(response.text)
        tables = pd.read_html(html_io)

        if not tables:
            raise ValueError(f"No tables found for year {year}.")

        return tables[0]

    except requests.exceptions.RequestException as e:
        print(f"Request failed: {e}")
    except ValueError as ve:
        print(f"Data error: {ve}")
    except Exception as ex:
        print(f"Unexpected error: {ex}")

    return pd.DataFrame()
```

```python
# "stacking dataframes"
# pd.concat() is used to concatenate the DataFrames vertically.
# The ignore_index=True parameter ensures that the resulting DataFrame has a new index,
# ignoring the original indices of the input DataFrames.
# The stacked_df now contains the concatenated DataFrame.
stacked_ipos_df = pd.concat([ipos_2026 ,ipos_2025, ipos_2024, ipos_2023], ignore_index=True)
```

```python
# Chaining: select one column, draw a plot, of a type line
nvo_df_filtered_from_2020['Close'].plot.line(title='Novo Nordisk A/S (NVO) price daily')
```

```python
# generating new fields (using DateTime features):

nvo_df['Ticker'] = 'NVO'
nvo_df['Year']= nvo_df.index.year
nvo_df['Month'] = nvo_df.index.month
nvo_df['Weekday'] = nvo_df.index.weekday
nvo_df['Date'] = nvo_df.index.date # to be used in joins
```

```python
nvo_df.tail()
```

```python
# shift ALL values (on x periods forward (+1) and backward (-1))
# equivalent of joining with a dataframe of the same vector, but with shifted date index
nvo_df['close_minus_1'] = nvo_df['Close'].shift(-1)
nvo_df['close_plus_1'] = nvo_df['Close'].shift(1)

nvo_df.tail()
```

```python
# historical growth
nvo_df['growth_1d'] = nvo_df['Close'] / nvo_df['Close'].shift(1)           # nvo_df['close_plus_1']
nvo_df['growth_30d'] = nvo_df['Close'] / nvo_df['Close'].shift(30)

# FUTURE Growth : for regression models
nvo_df['growth_future_1d'] = nvo_df['Close'].shift(-1) / nvo_df['Close']   # nvo_df['close_minus_1']
nvo_df['growth_future_30d'] = nvo_df['Close'].shift(-30) / nvo_df['Close']   # nvo_df['close_minus_1']


# FUTURE Growth: for binary models
nvo_df['is_positive_growth_1d_future'] = np.where(nvo_df['growth_future_1d'] > 1, 1, 0)
nvo_df['is_positive_growth_30d_future'] = np.where(nvo_df['growth_future_30d'] > 1, 1, 0)
```

```python
# normally the growth in 1 day is +-10%, while a lot of it is around 0% (around 1.)
plt.figure(figsize=(10, 6))
plt.title('Distribution of Daily/Monthly Growth Rates for ticker ="NVO"')

nvo_df.growth_1d.hist(bins=200, alpha=0.6, density=True)
nvo_df.growth_30d.hist(bins=200, alpha=0.6, density=True)

# Add vertical lines for averages
mean_1d = nvo_df.growth_1d.mean()
mean_30d = nvo_df.growth_30d.mean()

plt.axvline(mean_1d, color='blue', linestyle='--', linewidth=2, label=f'1d avg = {mean_1d:.3f}')
plt.axvline(mean_30d, color='orange', linestyle='--', linewidth=2, label=f'30d avg = {mean_30d:.3f}')


# Add mean text labels (as percent change)
ymax = plt.ylim()[1]
plt.text(mean_1d, ymax * 0.90,
         f'{(mean_1d - 1) * 100:.2f}%',
         color='blue', ha='center', va='bottom',
         bbox=dict(facecolor='white', edgecolor='blue', boxstyle='round,pad=0.2'))

plt.text(mean_30d, ymax * 0.85,
         f'{(mean_30d - 1) * 100:.2f}%',
         color='orange', ha='center', va='bottom',
         bbox=dict(facecolor='white', edgecolor='orange', boxstyle='round,pad=0.2'))


plt.show()
```

```python
# you can see that the growth is can be much wider in 30 days (+- 30%)
plt.title('Distribution of 2-days Growth Rates for ticker = "NVO"')

nvo_df.growth_30d.hist(bins=200)
```

```python
import re

def extract_numbers(input_string):
    y_match = re.search(r'Y(\d+)', input_string)
    m_match = re.search(r'M(\d+)', input_string)

    y_number = int(y_match.group(1)) if y_match else 0
    m_number = int(m_match.group(1)) if m_match else 0


    return y_number*12 + m_number

# Examples
examples = ["Y10_M2", "M3", "Y1"]
for example in examples:
    res = extract_numbers(example)
    print(f"{example} ==> {res}")
```

```python
def talib_get_momentum_indicators_for_one_ticker(df: pd.DataFrame) -> pd.DataFrame:
  # ADX - Average Directional Movement Index
  talib_momentum_adx = talib.ADX(df.High.values, df.Low.values, df.Close.values, timeperiod=14)
  # ADXR - Average Directional Movement Index Rating
  talib_momentum_adxr = talib.ADXR(df.High.values, df.Low.values, df.Close.values, timeperiod=14 )
  # APO - Absolute Price Oscillator
  talib_momentum_apo = talib.APO(df.Close.values, fastperiod=12, slowperiod=26, matype=0 )
  # AROON - Aroon
  talib_momentum_aroon = talib.AROON(df.High.values, df.Low.values, timeperiod=14 )
  # talib_momentum_aroon[0].size
  # talib_momentum_aroon[1].size
  # AROONOSC - Aroon Oscillator
  talib_momentum_aroonosc = talib.AROONOSC(df.High.values, df.Low.values, timeperiod=14)
  # BOP - Balance of Power
  # https://school.stockcharts.com/doku.php?id=technical_indicators:balance_of_power
     #calculate open prices as shifted closed prices from the prev day
     # open = df.Last.shift(1)
  talib_momentum_bop = talib.BOP(df.Open.values, df.High.values, df.Low.values, df.Close.values)
  # CCI - Commodity Channel Index
  talib_momentum_cci = talib.CCI(df.High.values, df.Low.values, df.Close.values, timeperiod=14)
  # CMO - Chande Momentum Oscillator
  talib_momentum_cmo = talib.CMO(df.Close.values, timeperiod=14)
  # DX - Directional Movement Index
  talib_momentum_dx = talib.DX(df.High.values, df.Low.values, df.Close.values, timeperiod=14)
  # MACD - Moving Average Convergence/Divergence
  talib_momentum_macd, talib_momentum_macdsignal, talib_momentum_macdhist = talib.MACD(df.Close.values, fastperiod=12, \
                                                                                       slowperiod=26, signalperiod=9)
  # MACDEXT - MACD with controllable MA type
  talib_momentum_macd_ext, talib_momentum_macdsignal_ext, talib_momentum_macdhist_ext = talib.MACDEXT(df.Close.values, \
                                                                                                    fastperiod=12, \
                                                                                                    fastmatype=0, \
                                                                                                    slowperiod=26, \
                                                                                                    slowmatype=0, \
                                                                                                    signalperiod=9, \
                                                                                                  signalmatype=0)
  # MACDFIX - Moving Average Convergence/Divergence Fix 12/26
  talib_momentum_macd_fix, talib_momentum_macdsignal_fix, talib_momentum_macdhist_fix = talib.MACDFIX(df.Close.values, \
                                                                                                      signalperiod=9)
  # MFI - Money Flow Index
  talib_momentum_mfi = talib.MFI(df.High.values, df.Low.values, df.Close.values, df.Volume.values, timeperiod=14)
  # MINUS_DI - Minus Directional Indicator
  talib_momentum_minus_di = talib.MINUS_DM(df.High.values, df.Low.values, timeperiod=14)
  # MOM - Momentum
  talib_momentum_mom = talib.MOM(df.Close.values, timeperiod=10)
  # PLUS_DI - Plus Directional Indicator
  talib_momentum_plus_di = talib.PLUS_DI(df.High.values, df.Low.values, df.Close.values, timeperiod=14)
  # PLUS_DM - Plus Directional Movement
  talib_momentum_plus_dm = talib.PLUS_DM(df.High.values, df.Low.values, timeperiod=14)
  # PPO - Percentage Price Oscillator
  talib_momentum_ppo = talib.PPO(df.Close.values, fastperiod=12, slowperiod=26, matype=0)
  # ROC - Rate of change : ((price/prevPrice)-1)*100
  talib_momentum_roc = talib.ROC(df.Close.values, timeperiod=10)
  # ROCP - Rate of change Percentage: (price-prevPrice)/prevPrice
  talib_momentum_rocp = talib.ROCP(df.Close.values, timeperiod=10)
  # ROCR - Rate of change ratio: (price/prevPrice)
  talib_momentum_rocr = talib.ROCR(df.Close.values, timeperiod=10)
  # ROCR100 - Rate of change ratio 100 scale: (price/prevPrice)*100
  talib_momentum_rocr100 = talib.ROCR100(df.Close.values, timeperiod=10)
  # RSI - Relative Strength Index
  talib_momentum_rsi = talib.RSI(df.Close.values, timeperiod=14)
  # STOCH - Stochastic
  talib_momentum_slowk, talib_momentum_slowd = talib.STOCH(df.High.values, df.Low.values, df.Close.values, \
                                                           fastk_period=5, slowk_period=3, slowk_matype=0, slowd_period=3, slowd_matype=0)
  # STOCHF - Stochastic Fast
  talib_momentum_fastk, talib_momentum_fastd = talib.STOCHF(df.High.values, df.Low.values, df.Close.values, \
                                                            fastk_period=5, fastd_period=3, fastd_matype=0)
  # STOCHRSI - Stochastic Relative Strength Index
  talib_momentum_fastk_rsi, talib_momentum_fastd_rsi = talib.STOCHRSI(df.Close.values, timeperiod=14, \
                                                                      fastk_period=5, fastd_period=3, fastd_matype=0)
  # TRIX - 1-day Rate-Of-Change (ROC) of a Triple Smooth EMA
  talib_momentum_trix = talib.TRIX(df.Close.values, timeperiod=30)
  # ULTOSC - Ultimate Oscillator
  talib_momentum_ultosc = talib.ULTOSC(df.High.values, df.Low.values, df.Close.values, timeperiod1=7, timeperiod2=14, timeperiod3=28)
  # WILLR - Williams' %R
  talib_momentum_willr = talib.WILLR(df.High.values, df.Low.values, df.Close.values, timeperiod=14)

  momentum_df =   pd.DataFrame(
    {
      # assume here multi-index <dateTime, ticker>
      # 'datetime': df.index.get_level_values(0),
      # 'ticker': df.index.get_level_values(1) ,

      # old way with separate columns
      'Date': df.Date.values,
      'Ticker': df.Ticker,

      'adx': talib_momentum_adx,
      'adxr': talib_momentum_adxr,
      'apo': talib_momentum_apo,
      'aroon_1': talib_momentum_aroon[0] ,
      'aroon_2': talib_momentum_aroon[1],
      'aroonosc': talib_momentum_aroonosc,
      'bop': talib_momentum_bop,
      'cci': talib_momentum_cci,
      'cmo': talib_momentum_cmo,
      'dx': talib_momentum_dx,
      'macd': talib_momentum_macd,
      'macdsignal': talib_momentum_macdsignal,
      'macdhist': talib_momentum_macdhist,
      'macd_ext': talib_momentum_macd_ext,
      'macdsignal_ext': talib_momentum_macdsignal_ext,
      'macdhist_ext': talib_momentum_macdhist_ext,
      'macd_fix': talib_momentum_macd_fix,
      'macdsignal_fix': talib_momentum_macdsignal_fix,
      'macdhist_fix': talib_momentum_macdhist_fix,
      'mfi': talib_momentum_mfi,
      'minus_di': talib_momentum_minus_di,
      'mom': talib_momentum_mom,
      'plus_di': talib_momentum_plus_di,
      'dm': talib_momentum_plus_dm,
      'ppo': talib_momentum_ppo,
      'roc': talib_momentum_roc,
      'rocp': talib_momentum_rocp,
      'rocr': talib_momentum_rocr,
      'rocr100': talib_momentum_rocr100,
      'rsi': talib_momentum_rsi,
      'slowk': talib_momentum_slowk,
      'slowd': talib_momentum_slowd,
      'fastk': talib_momentum_fastk,
      'fastd': talib_momentum_fastd,
      'fastk_rsi': talib_momentum_fastk_rsi,
      'fastd_rsi': talib_momentum_fastd_rsi,
      'trix': talib_momentum_trix,
      'ultosc': talib_momentum_ultosc,
      'willr': talib_momentum_willr,
     }
  )
  return momentum_df
```

## 6.2 Volume, Volatility, Cycle, Price indicators
https://github.com/TA-Lib/ta-lib-python/blob/master/docs/func_groups/volume_indicators.md

```python
def talib_get_volume_volatility_cycle_price_indicators(df: pd.DataFrame) -> pd.DataFrame:
        # TA-Lib Volume indicators
        # https://github.com/TA-Lib/ta-lib-python/blob/master/docs/func_groups/volume_indicators.md
        # AD - Chaikin A/D Line
        talib_ad = talib.AD(
            df.High.values, df.Low.values, df.Close.values, df.Volume.values)
        # ADOSC - Chaikin A/D Oscillator
        talib_adosc = talib.ADOSC(
            df.High.values, df.Low.values, df.Close.values, df.Volume.values, fastperiod=3, slowperiod=10)
        # OBV - On Balance Volume
        talib_obv = talib.OBV(
            df.Close.values, df.Volume.values)

        # TA-Lib Volatility indicators
        # https://github.com/TA-Lib/ta-lib-python/blob/master/docs/func_groups/volatility_indicators.md
        # ATR - Average True Range
        talib_atr = talib.ATR(
            df.High.values, df.Low.values, df.Close.values, timeperiod=14)
        # NATR - Normalized Average True Range
        talib_natr = talib.NATR(
            df.High.values, df.Low.values, df.Close.values, timeperiod=14)
        # OBV - On Balance Volume
        talib_obv = talib.OBV(
            df.Close.values, df.Volume.values)

        # TA-Lib Cycle Indicators
        # https://github.com/TA-Lib/ta-lib-python/blob/master/docs/func_groups/cycle_indicators.md
        # HT_DCPERIOD - Hilbert Transform - Dominant Cycle Period
        talib_ht_dcperiod = talib.HT_DCPERIOD(df.Close.values)
        # HT_DCPHASE - Hilbert Transform - Dominant Cycle Phase
        talib_ht_dcphase = talib.HT_DCPHASE(df.Close.values)
        # HT_PHASOR - Hilbert Transform - Phasor Components
        talib_ht_phasor_inphase, talib_ht_phasor_quadrature = talib.HT_PHASOR(
            df.Close.values)
        # HT_SINE - Hilbert Transform - SineWave
        talib_ht_sine_sine, talib_ht_sine_leadsine = talib.HT_SINE(
            df.Close.values)
        # HT_TRENDMODE - Hilbert Transform - Trend vs Cycle Mode
        talib_ht_trendmode = talib.HT_TRENDMODE(df.Close.values)

        # TA-Lib Price Transform Functions
        # https://github.com/TA-Lib/ta-lib-python/blob/master/docs/func_groups/price_transform.md
        # AVGPRICE - Average Price
        talib_avgprice = talib.AVGPRICE(
            df.Open.values, df.High.values, df.Low.values, df.Close.values)
        # MEDPRICE - Median Price
        talib_medprice = talib.MEDPRICE(df.High.values, df.Low.values)
        # TYPPRICE - Typical Price
        talib_typprice = talib.TYPPRICE(
            df.High.values, df.Low.values, df.Close.values)
        # WCLPRICE - Weighted Close Price
        talib_wclprice = talib.WCLPRICE(
            df.High.values, df.Low.values, df.Close.values)

        volume_volatility_cycle_price_df = pd.DataFrame(
            {'Date': df.Date.values,
             'Ticker': df.Ticker,
             # TA-Lib Volume indicators
             'ad': talib_ad,
             'adosc': talib_adosc,
             'obv': talib_obv,
             # TA-Lib Volatility indicators
             'atr': talib_atr,
             'natr': talib_natr,
             'obv': talib_obv,
             # TA-Lib Cycle Indicators
             'ht_dcperiod': talib_ht_dcperiod,
             'ht_dcphase': talib_ht_dcphase,
             'ht_phasor_inphase': talib_ht_phasor_inphase,
             'ht_phasor_quadrature': talib_ht_phasor_quadrature,
             'ht_sine_sine': talib_ht_sine_sine,
             'ht_sine_leadsine': talib_ht_sine_leadsine,
             'ht_trendmod': talib_ht_trendmode,
             # TA-Lib Price Transform Functions
             'avgprice': talib_avgprice,
             'medprice': talib_medprice,
             'typprice': talib_typprice,
             'wclprice': talib_wclprice,
             }
        )

        # Need a proper date type
        volume_volatility_cycle_price_df['Date'] = pd.to_datetime(
            volume_volatility_cycle_price_df['Date'])

        return volume_volatility_cycle_price_df
```

###1. What is the total withdrawn IPO value (in $ millions) for the company class with the highest total withdrawal value? (2 points)

```python
# =============================================================================
# Q1 STEP 1: Load the "Recently Filed IPOs" table from iposcoop.com
# 목표: 웹페이지의 HTML 표를 그대로 DataFrame으로 가져오기
# =============================================================================
import io          # StringIO: HTML 문자열을 버퍼로 감싸기 위해 필요
import requests    # HTTP 요청
import numpy as np
import pandas as pd

URL = "https://www.iposcoop.com/ipos-recently-filed/"

# User-Agent header: pretend to be a real browser.
# 이게 없으면 많은 사이트가 파이썬 봇으로 판단해 403 Forbidden을 반환함
HEADERS = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 '
                         '(KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'}

resp = requests.get(URL, headers=HEADERS, timeout=20)
resp.raise_for_status()   # 404/500 등을 예외로 승격 → 에러 페이지를 데이터로 착각하는 사고 방지

# io.StringIO 로 감싸는 이유 (숙제 안내문의 함정 #2):
#   pd.read_html(resp.text) 를 직접 호출하면
#   Windows + 최신 pandas/lxml 조합에서 OSError 발생 가능
tables = pd.read_html(io.StringIO(resp.text))

# read_html returns a LIST of every <table> on the page.
# 사이드바·광고 표가 섞일 수 있으므로 항상 개수와 모양을 먼저 확인할 것
print(f"페이지에서 찾은 테이블 개수: {len(tables)}")

df = tables[0].copy()   # .copy() → 이후 컬럼 추가 시 SettingWithCopyWarning 방지

# Normalize column names: strip whitespace and newlines from <th> text
# HTML 헤더에는 줄바꿈/앞뒤 공백이 섞여 있는 경우가 흔함
df.columns = [str(c).replace('\n', ' ').strip() for c in df.columns]

print(f"shape: {df.shape}")
print(list(df.columns))
df.head(3)
```

```python
# =============================================================================
# Q1 STEP 2: Keep only rows where 'Expected To Trade' == 'Withdrawn'
# 체크포인트: 32건 근처여야 정상 (숙제 안내문: 몇 건 차이는 정상)
# =============================================================================

# 먼저 이 컬럼에 어떤 값들이 존재하는지 확인 → 필터 조건을 정확히 짜기 위해
# dropna=False 를 주면 NaN 개수도 함께 표시됨
print(df['Expected To Trade'].value_counts(dropna=False), '\n')

withdrawn = df[
    df['Expected To Trade']
      .astype(str)      # NaN 이 섞여 있어도 .str 접근자가 죽지 않도록 방어
      .str.strip()      # HTML 셀의 앞뒤 공백 제거 ('Withdrawn ' != 'Withdrawn')
      == 'Withdrawn'
].copy()                # 슬라이스가 아닌 독립 복사본으로 → 이후 컬럼 추가 안전

print("철회(Withdrawn) 건수:", len(withdrawn))   # ← 32 근처 기대: 34!
```

```python
# =============================================================================
# Q1 STEP 3: Create 'Company Type' by pattern-matching the company name
# 핵심 규칙: ORDER MATTERS — 첫 번째로 매칭되는 규칙을 적용
# =============================================================================

def classify_company(name) -> str:
    """
    Classify a company by name patterns. First match wins.

    if/elif 대신 각 분기에서 곧바로 return 하므로,
    첫 매칭 시점에 함수가 종료됨 → "첫 매칭 우선" 규칙이 구조적으로 보장됨
    """
    s = str(name)   # NaN 등 비문자열 입력 방어

    # 주의: 'Technologies' 는 복수형 정확 매칭.
    #       'Technology'(단수)는 여기에 걸리지 않음 → 숙제 검증 사례 #2
    if 'Technologies' in s:
        return 'Technologies'

    # 'Corp' 이 'Acquisition Corp' 를 이미 포함하므로 셋 다 같은 결과.
    # 숙제 문서 표현을 그대로 옮긴 형태
    if ('Acquisition Corp' in s) or ('Acquisition Corporation' in s) or ('Corp' in s):
        return 'Acquisition Corp'

    if ('Inc' in s) or ('Incorporated' in s):
        return 'Inc.'

    # 이 규칙이 Ltd/Limited 보다 위에 있어야 함 → 숙제 검증 사례 #1
    if 'Group' in s:
        return 'Group'

    if ('Ltd' in s) or ('Limited' in s):
        return 'Limited'

    if ('Holdings' in s) or ('Holding' in s):
        return 'Holdings'

    return 'Other'   # 어느 규칙에도 안 걸리면 기타


# ---- Sanity checks: 숙제 문서가 직접 제시한 두 가지 사례 ----
# "Group" 규칙이 "Ltd" 규칙보다 먼저 오므로 Group 으로 분류되어야 함
assert classify_company("EUPEC International Group Ltd.") == 'Group'
# "Technology"(단수)는 "Technologies"(복수)와 매칭되지 않으므로 Limited 가 되어야 함
assert classify_company("Xinxu Copper Industry Technology Ltd.") == 'Limited'
print("분류 규칙 검증 통과 ✓\n")

# .apply() = 행마다 함수를 적용 (모듈2 §5의 get_ticker_type 패턴과 동일)
withdrawn['Company Type'] = withdrawn['Company'].apply(classify_company)

print(withdrawn['Company Type'].value_counts(), '\n')

# 34행을 전부 눈으로 확인 — 'Other' 가 과도하게 많으면 규칙에 빠진 패턴이 있다는 신호
# .to_string() 은 반환값이므로 print() 로 감싸야 출력됨
print(withdrawn[['Company', 'Company Type']].to_string())
```

```python
# =============================================================================
# Q1 STEP 4~5: Parse numeric fields, then build 'Shares_offered_value'
#   - Avg_price            = (Price Low + Price High) / 2
#   - Shares_offered_value = shares * Avg_price, 없으면 Est $ Vol 로 대체
# =============================================================================

def to_num(series):
    """
    Clean a currency/number column into float.
        '$8.00'  -> 8.0
        '1,234'  -> 1234.0
        '-'      -> NaN
        ''/None  -> NaN

    regex=True 로 '$' 와 ',' 를 한 번에 제거한 뒤 숫자로 변환.
    errors='coerce' → 파싱 실패 시 예외 대신 NaN (모듈2 §1의 '방어적 변환' 패턴)
    """
    return pd.to_numeric(
        series.astype(str).str.replace(r'[$,]', '', regex=True).str.strip(),
        errors='coerce')


# ---- Price parsing ----
withdrawn['price_low_num']  = to_num(withdrawn['Price Low'])
withdrawn['price_high_num'] = to_num(withdrawn['Price High'])

# .mean(axis=1) 을 쓰는 이유:
#   (low + high) / 2 는 한쪽이 NaN 이면 결과 전체가 NaN
#   .mean(axis=1) 은 NaN 을 건너뛰고 존재하는 값만으로 평균을 냄
withdrawn['Avg_price'] = withdrawn[['price_low_num', 'price_high_num']].mean(axis=1)

# ---- Shares & estimated volume ----
# 두 컬럼 모두 단위가 "millions" 이므로 곱셈 결과도 백만 달러 단위로 일치함
withdrawn['shares_num']  = to_num(withdrawn['Shares (millions)'])
withdrawn['est_vol_num'] = to_num(withdrawn['Est $ Vol (millions)'])

# ---- Final value: COALESCE(계산값, Est $ Vol) ----
# .fillna() 가 곧 "우선순위 대체" 의미론 (SQL 의 COALESCE 와 동일)
withdrawn['Shares_offered_value'] = (
    (withdrawn['shares_num'] * withdrawn['Avg_price'])   # 1순위: 주식수 × 평균가
    .fillna(withdrawn['est_vol_num'])                    # 2순위: 사이트가 제시한 예상 규모
)

# ---- 검증 출력 ----
# 원본 문자열과 파싱 결과를 나란히 놓고 눈으로 대조
print(withdrawn[['Company', 'Price Low', 'Price High', 'Avg_price',
                 'shares_num', 'est_vol_num', 'Shares_offered_value']].to_string())

# 결측이 몇 개인지 확인:
#   몇 건은 정상(둘 다 값이 없는 행)이지만,
#   32건 전부 NaN 이면 컬럼명이나 파싱이 잘못된 것
print("\n결측 개수:")
print(withdrawn[['Avg_price', 'shares_num', 'est_vol_num', 'Shares_offered_value']].isna().sum())

# 대체 로직이 실제로 몇 번 작동했는지 확인 (철회 IPO는 주식수 미확정이 흔함)
n_fallback = withdrawn['shares_num'].mul(withdrawn['Avg_price']).isna().sum()
print(f"\nEst $ Vol 로 대체된 행 수: {n_fallback}")
```

```python
# =============================================================================
# Q1 STEP 6: Group by Company Type and sum Shares_offered_value
# 정답 = 합계가 가장 큰 유형 + 그 합계 값
# =============================================================================

# sum 외에 count/mean 도 함께 보는 이유:
#   건수 1건인데 합계 1위라면 → 거대 단일 건이 지배하는 것
#   즉 분류나 파싱에 문제가 있을 수 있다는 신호 (검산용)
# .sum() 은 NaN 을 자동으로 무시하므로 결측 행은 제외됨
result = (withdrawn
          .groupby('Company Type')['Shares_offered_value']
          .agg(['sum', 'count', 'mean'])
          .sort_values('sum', ascending=False))

print(result, '\n')
print(f"정답: {result.index[0]} — ${result.iloc[0]['sum']:.1f}M")
```

```python
# =============================================================================
# Q2 STEP 1: Load the list of 2025 IPOs from iposcoop.com
# 목표: 2025년에 가격이 확정된(=실제 상장된) IPO 231건의 표를 가져오기
# 참고: Q1은 'ipos-recently-filed'(신청 단계), 여기는 '2025-pricings'(체결 단계) — 다른 페이지
# =============================================================================
import io
import time
import requests
import numpy as np
import pandas as pd
import yfinance as yf

URL_2025 = "https://www.iposcoop.com/2025-pricings/"
HEADERS = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 '
                         '(KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'}

resp = requests.get(URL_2025, headers=HEADERS, timeout=20)
resp.raise_for_status()

# Q1과 동일하게 io.StringIO 로 감싸기 (Windows/최신 pandas 의 OSError 회피)
tables = pd.read_html(io.StringIO(resp.text))
print(f"페이지에서 찾은 테이블 개수: {len(tables)}")

ipos = tables[0].copy()
ipos.columns = [str(c).replace('\n', ' ').strip() for c in ipos.columns]

print(f"shape: {ipos.shape}")        # ← 231행 근처 기대
print(list(ipos.columns))            # ← 실제 컬럼명 확인 (아래 셀에서 사용)
ipos.head(5)
```

```python
# =============================================================================
# Q2 STEP 2: Keep IPOs that (a) priced before 2025-09-01, (b) have non-zero Return
# 체크포인트: 148종목 근처
#
# 왜 이 두 필터인가:
#   (a) 2026-09-11 시점에 "1년(252거래일) 데이터"가 쌓이려면 상장이 충분히 일렀어야 함
#   (b) Return 이 정확히 0% = 야후에 데이터가 없거나 거래가 없는 비활성 티커일 가능성 → 제외
# =============================================================================

date_col   = 'Offer Date'    # ← 셀 1 출력에 맞춰 수정
symbol_col = 'Symbol'        # ← 셀 1 출력에 맞춰 수정
return_col = 'Return'        # ← 셀 1 출력에 맞춰 수정

# ---- (1) 날짜 파싱 ----
# format='mixed' : 행마다 날짜 표기가 달라도 각각 추론 (pandas 2.0+)
# errors='coerce': 파싱 실패 시 예외 대신 NaT → 파이프라인이 멈추지 않음
ipos['offer_date'] = pd.to_datetime(ipos[date_col], format='mixed', errors='coerce')
print("날짜 파싱 실패(NaT) 건수:", ipos['offer_date'].isna().sum())

# ---- (2) Return 을 숫자로 ----
# '-12.50%' -> -12.5  (퍼센트 기호 제거만, /100 은 생략 — 0 비교만 할 것이므로 무관)
ipos['return_num'] = pd.to_numeric(
    ipos[return_col].astype(str).str.replace('%', '', regex=False).str.strip(),
    errors='coerce')

# ---- (3) 두 조건 결합 ----
# pandas 에서는 and/or 대신 &/| 를 쓰고, 각 조건을 괄호로 감싸야 함 (연산자 우선순위 때문)
mask = (ipos['offer_date'] < '2025-09-01') & (ipos['return_num'] != 0)
filtered = ipos[mask].copy()

# ---- (4) 티커 정리 ----
# HTML 셀에 공백이나 대소문자 혼용이 있을 수 있으므로 정규화
filtered['ticker'] = filtered[symbol_col].astype(str).str.strip().str.upper()

# 중복 티커 제거 (같은 회사가 두 번 실릴 수 있음)
tickers = filtered['ticker'].dropna().unique().tolist()
tickers = [t for t in tickers if t not in ('', 'NAN', '-')]   # 빈 값 방어

print(f"필터 후 종목 수: {len(tickers)}")   # ← 148 근처 기대
print(tickers[:20])
```

```python
# =============================================================================
# Q2 STEP 3: Download daily OHLCV for each ticker via yfinance
# 체크포인트: 성공 134종목 근처 (상장폐지 등으로 14개 내외가 실패하는 게 정상)
#
# 설계 원칙 (모듈2 §5와 동일):
#   - 종목 하나씩 받아서 Ticker 컬럼을 붙인 뒤 리스트에 모음
#   - 마지막에 한 번만 concat → 매 반복 concat 하는 O(n²) 회피
#   - 실패 종목은 건너뛰되 기록해서 나중에 확인
# =============================================================================

frames = []          # 성공한 DataFrame 들을 모을 리스트
failed  = []         # 실패한 티커 기록

for i, ticker in enumerate(tickers, start=1):
    try:
        # .history() 를 쓰는 이유 (숙제 안내문의 함정 #1 회피):
        #   yf.download() 는 최신 버전에서 MultiIndex 컬럼을 반환해
        #   df['Close'] 가 Series 대신 DataFrame 이 되는 문제가 있음.
        #   .history() 는 단일 종목 + 평평한 컬럼 + DatetimeIndex 를 보장
        h = yf.Ticker(ticker).history(period="max", interval="1d")

        # 상장폐지/미존재 종목은 예외가 아니라 '빈 DataFrame'으로 돌아오는 경우가 더 흔함
        # → .empty 체크가 try/except 보다 실질적으로 더 중요
        if h.empty:
            failed.append(ticker)
            continue

        h = h.copy()
        h['Ticker'] = ticker
        # 타임존이 붙은 DatetimeIndex → 순수 날짜 컬럼으로 (이후 비교/merge 를 위해)
        h['Date'] = h.index.tz_localize(None).normalize()

        frames.append(h)

    except Exception as e:
        failed.append(ticker)

    # Yahoo 서버 과부하 방지 — 없으면 rate limit 에 걸려 빈 응답이 돌아옴
    time.sleep(1)

    if i % 25 == 0:
        print(f"  진행 {i}/{len(tickers)} ... 성공 {len(frames)}")

# 한 번에 concat. ignore_index=True → 인덱스 중복 방지
# (대신 DatetimeIndex 가 사라지므로 위에서 'Date' 컬럼을 미리 만들어 둔 것)
stocks_df = pd.concat(frames, ignore_index=True)

print(f"\n성공: {stocks_df['Ticker'].nunique()} 종목")   # ← 134 근처 기대
print(f"실패: {len(failed)} 종목 → {failed}")
print(f"전체 행 수: {len(stocks_df):,}")
```

```python
# =============================================================================
# Q2 STEP 4~5: growth_252d, volatility, Sharpe 계산
#
# ⚠️ 핵심 주의점:
#   stocks_df 는 132개 종목이 세로로 쌓인 형태 (68,083행).
#   shift() / rolling() 을 그냥 호출하면
#   A종목의 끝부분에 B종목의 앞부분 값이 섞여 들어감!
#   → 반드시 groupby('Ticker') 안에서 계산해야 함
# =============================================================================

# 정렬은 순차 연산(shift/rolling)의 전제조건.
# 종목별로 날짜 오름차순이어야 "252행 전 = 252거래일 전"이 성립함
stocks_df = stocks_df.sort_values(['Ticker', 'Date']).reset_index(drop=True)

# ---- growth_252d : 약 1년 전 대비 성장률 ----
# 252 = 1년치 거래일 수 (365일 × 5/7 − 공휴일)
# 비율로 유지: 1.05 = +5%  (-1 을 하지 않음 — 숙제 공식이 비율 기준)
# transform() 은 groupby 결과를 원본과 같은 길이로 돌려줌 → 컬럼에 바로 대입 가능
stocks_df['growth_252d'] = (
    stocks_df.groupby('Ticker')['Close']
             .transform(lambda s: s / s.shift(252))
)

# ---- volatility : 연율화 변동성 ----
# ⚠️ 숙제가 지정한 공식을 그대로 사용:
#       stocks_df['Close'].rolling(30).std() * np.sqrt(252)
#
#   금융 교과서상으로는 '수익률'의 표준편차를 써야 맞고,
#   강의 노트북도 2025년에 growth_1d 기준으로 수정했음(§5 셀 128 주석 참고).
#   하지만 숙제 정답 보기가 Close 기준으로 산출되어 있으므로 여기서는 숙제를 따름.
#
#   sqrt(252) = 일간 → 연간 환산 계수
#   (분산은 시간에 비례하므로 표준편차는 시간의 제곱근에 비례)
stocks_df['volatility'] = (
    stocks_df.groupby('Ticker')['Close']
             .transform(lambda s: s.rolling(30).std() * np.sqrt(252))
)

# ---- Sharpe ratio ----
# (수익률 − 무위험수익률) / 변동성
# 0.05 = 미 10년물 국채 수익률 근사 — "가만히 있어도 얻는 수익"을 차감해
#        위험을 감수한 대가만 남김
stocks_df['Sharpe'] = (stocks_df['growth_252d'] - 0.05) / stocks_df['volatility']

# ---- 검증: 계산이 제대로 됐는지 확인 ----
print("컬럼별 non-null 개수:")
print(stocks_df[['growth_252d', 'volatility', 'Sharpe']].notna().sum(), '\n')

# groupby 가 제대로 작동했는지 확인용:
#   각 종목의 첫 252행은 growth_252d 가 NaN 이어야 정상
#   (만약 종목 경계에서 값이 이어져 있다면 groupby 를 빠뜨린 것)
check = (stocks_df.groupby('Ticker')['growth_252d']
                  .apply(lambda s: s.notna().sum()))
print("종목별 growth_252d 유효 행 수 (상위 5개):")
print(check.sort_values(ascending=False).head(), '\n')
print(f"growth_252d 가 하나도 없는 종목 수: {(check == 0).sum()} "
      f"(= 상장 1년 미만 종목)")

stocks_df[['Ticker','Date','Close','growth_252d','volatility','Sharpe']].tail(5)
```

```python
# =============================================================================
# Q2 STEP 6: 2026-09-11 하루만 잘라내어 describe()
# 정답 = Sharpe 의 50% (중앙값)
# =============================================================================

TARGET_DATE = pd.Timestamp('2026-09-11')

snapshot = stocks_df[stocks_df['Date'] == TARGET_DATE].copy()
print(f"{TARGET_DATE.date()} 에 거래된 종목 수: {len(snapshot)}\n")

# 0 이 나오면 그날이 휴장일이거나 Date 타입이 어긋난 것 → 주변 거래일 확인
if len(snapshot) == 0:
    nearby = stocks_df.loc[stocks_df['Date'].between('2026-09-08','2026-09-16'), 'Date']
    print("⚠️ 해당일 데이터 없음. 주변 거래일:", sorted(nearby.unique()))

# describe() = count/mean/std/min/25%/50%/75%/max
# 이 표의 'Sharpe' 열 '50%' 행이 정답
stats = snapshot[['growth_252d', 'volatility', 'Sharpe']].describe()
print(stats, '\n')

# count 는 "252거래일을 채운 종목 수"를 의미:
#   2025년 하반기 상장 종목은 아직 1년이 안 돼 growth_252d 가 NaN → 자동 제외됨
n_reached = int(snapshot['growth_252d'].notna().sum())
print(f"252일 마일스톤 도달 종목 수: {n_reached} / {len(snapshot)}\n")

median_sharpe = snapshot['Sharpe'].median()
median_growth = snapshot['growth_252d'].median()
mean_growth   = snapshot['growth_252d'].mean()

print(f"▶ 정답 — 중앙값 Sharpe: {median_sharpe:.4f}")
print(f"   참고 — growth_252d 중앙값 {median_growth:.4f} / 평균 {mean_growth:.4f}")
```

```python
# =============================================================================
# 숙제가 요구한 관찰 3가지
# =============================================================================

# 1) 평균 vs 중앙값 — 고성장 이상치가 평균을 편향시키는가?
gap = mean_growth - median_growth
print(f"1) growth_252d 평균 − 중앙값 = {gap:+.4f}")
print(f"   → {'소수의 폭등 종목이 평균을 끌어올림 (우측 꼬리)' if gap > 0 else '평균이 더 보수적'}\n")

# 2) 252일 도달 종목 수 (셀 5 출력 참고)

# 3) 위험조정 성과가 순수 성장률보다 매력적인 종목:
#    수익률은 평범해도 변동성이 낮아 Sharpe 가 높은 케이스
top = snapshot.nlargest(10, 'Sharpe')[['Ticker','growth_252d','volatility','Sharpe']]
print("3) Sharpe 상위 10종목:")
print(top.to_string(index=False))
```

###3. What is the optimal number of months (1 to 12) to hold a newly IPO'd stock in order to maximize the median growth value? (3 points)
####1 / 3 / 5 / 7

```python
# =============================================================================
# Q3 STEP 2: 1~12개월 보유 시 미래 성장률 컬럼 12개 생성
#
# 가정: 1개월 = 21 거래일  (1m=21, 2m=42, ..., 12m=252)
#       달력 날짜가 아니라 '행 개수' 기준 — shift() 는 행을 움직이므로
#       주말/공휴일은 애초에 행이 없어 자동으로 제외됨
#
# ⚠️ stocks_df 는 132종목이 쌓인 형태 → 반드시 groupby('Ticker')
# =============================================================================

# 순차 연산의 전제조건: 종목별 날짜 오름차순
stocks_df = stocks_df.sort_values(['Ticker', 'Date']).reset_index(drop=True)

TRADING_DAYS_PER_MONTH = 21

for m in range(1, 13):
    horizon = TRADING_DAYS_PER_MONTH * m          # 21, 42, 63, ... 252
    col = f'future_growth_{m}_m'

    # shift(-n) = 미래 값을 현재 행으로 끌어옴 (모듈2 §2.1 참고)
    #   future_growth = n거래일 뒤 종가 / 오늘 종가
    #   비율 유지: 1.15 = +15%
    stocks_df[col] = (
        stocks_df.groupby('Ticker')['Close']
                 .transform(lambda s, h=horizon: s.shift(-h) / s)
        #                          ^^^^^^^^^^ 기본값으로 h 를 고정(late binding 방지)
        #                          lambda 가 루프 변수를 참조하면 마지막 값만 쓰이는 버그 발생
    )

growth_cols = [f'future_growth_{m}_m' for m in range(1, 13)]

print("생성된 컬럼:", growth_cols, '\n')
print("컬럼별 non-null 개수 (뒤로 갈수록 줄어야 정상):")
print(stocks_df[growth_cols].notna().sum())
```

```python
# =============================================================================
# Q3 STEP 3~4: 각 티커의 첫 거래일(min_date)을 찾고, 그 행만 추출
#
# 목적: "IPO 당일에 종가로 사서 N개월 보유"의 수익률만 골라내기
#       전체 데이터를 그냥 평균내면 '아무 날이나 산 경우'가 섞여버림
# =============================================================================

# 각 종목의 데이터 시작일
min_dates = (stocks_df.groupby('Ticker')['Date']
                      .min()
                      .reset_index()
                      .rename(columns={'Date': 'min_date'}))

print(f"종목 수: {len(min_dates)}")
print(min_dates['min_date'].dt.year.value_counts().sort_index(), '\n')

# ---- Inner join: min_date 행만 남기기 ----
# 왼쪽(min_dates)은 종목당 1행 → 결과도 종목당 1행
# validate='one_to_one' 으로 행 폭증을 원천 차단 (모듈2 §6 패턴)
entry = min_dates.merge(
    stocks_df,
    left_on=['Ticker', 'min_date'],
    right_on=['Ticker', 'Date'],
    how='inner',
    validate='one_to_one'
)

print(f"진입 시점 행 수: {len(entry)}  (종목 수와 같아야 정상)\n")
print(entry[['Ticker', 'min_date', 'Close'] + growth_cols[:3]].head(10).to_string(index=False))
```

```python
# =============================================================================
# Q3 STEP 5: 12개 컬럼의 describe() → 50%(중앙값)가 가장 높은 달 찾기
# =============================================================================

stats = entry[growth_cols].describe()

# 소수점 4자리로 보기 좋게
pd.set_option('display.width', 200)
print(stats.round(4).to_string(), '\n')

# ---- 중앙값만 뽑아서 비교 ----
medians = entry[growth_cols].median()

print("월별 중앙값 (future_growth):")
for m, (col, val) in enumerate(medians.items(), start=1):
    bar = '█' * int(max(val, 0) * 30)          # 간단한 시각화
    n = int(entry[col].notna().sum())          # 유효 표본 수
    print(f"  {m:2d}개월: {val:.4f}  (n={n:3d})  {bar}")

best_col = medians.idxmax()
best_month = int(best_col.split('_')[2])
best_value = medians.max()

print(f"\n▶ 정답 — 최적 보유기간: {best_month}개월, 중앙값 성장률 {best_value:.4f}")
print(f"   (= {(best_value - 1) * 100:+.2f}% 수익)")
```

```python
# =============================================================================
# 숙제가 요구한 관찰 + 이상치 5종목의 영향 확인
# =============================================================================

# ---- 1) 중앙값의 시간 추세 ----
print("1) 중앙값 추세:")
trend = medians.values
print(f"   1개월 {trend[0]:.4f} → 12개월 {trend[-1]:.4f}")
print(f"   → {'시간이 갈수록 하락 (IPO 후 초기 버블이 빠지는 패턴)' if trend[-1] < trend[0] else '시간이 갈수록 상승'}\n")

# ---- 2) 중앙값 vs 평균 ----
print("2) 중앙값 vs 평균 (고성장 이상치의 영향):")
comp = pd.DataFrame({
    'median': entry[growth_cols].median(),
    'mean':   entry[growth_cols].mean(),
})
comp['gap'] = comp['mean'] - comp['median']
print(comp.round(4).to_string(), '\n')
print("   → gap 이 양수 = 소수의 폭등 종목이 평균을 끌어올림.")
print("     전형적인 투자자가 경험하는 건 median 쪽에 가까움\n")

# ---- 3) 2025년 이전 시작 5종목을 뺐을 때 답이 바뀌는지 ----
SUSPICIOUS = ['CAPS', 'AVBH', 'PPCB', 'ALM', 'CIIT']
entry_clean = entry[~entry['Ticker'].isin(SUSPICIOUS)]

medians_clean = entry_clean[growth_cols].median()
best_clean = medians_clean.idxmax()

print("3) 티커 재활용 의심 5종목 제외 시:")
print(f"   제외 전: {best_col} = {medians.max():.4f}")
print(f"   제외 후: {best_clean} = {medians_clean.max():.4f}")
print(f"   → 정답 {'동일 ✓' if best_clean == best_col else '변경됨 ⚠️'}")
```

###4. What is the total profit (in thousands dollars) you would have earned by investing 1000 dollars every time a stock was oversold (RSI < 30)? (2 points)

```python
# =============================================================================
# Q4 STEP 1: 미리 계산된 기술지표/거시지표가 담긴 parquet 파일 다운로드
#
# 이 파일의 정체 = 모듈2 §5+§6+§7 파이프라인의 최종 산출물 (m16)
#   §5 가격·성장률 → §6 TA-Lib 지표(rsi 포함) → §7 거시지표 merge
#   단, 33종목이 아니라 수백 종목 규모로 돌린 버전
# =============================================================================
!pip install gdown -q

import gdown
import pandas as pd
import numpy as np

file_id = "1grCTCzMZKY5sJRtdbLVCXg8JXA8VPyg-"
gdown.download(f"https://drive.google.com/uc?id={file_id}", "data.parquet", quiet=False)

# parquet 은 열 지향 포맷 → 필요한 컬럼만 읽을 수도 있지만,
# 일단 전체를 읽어 구조를 파악
df = pd.read_parquet("data.parquet", engine="pyarrow")

print(f"shape: {df.shape}")
print(f"기간: {df['Date'].min()} ~ {df['Date'].max()}")
print(f"종목 수: {df['Ticker'].nunique()}\n")

# 우리가 쓸 두 컬럼이 실제로 있는지 확인
print("rsi 관련 컬럼:", [c for c in df.columns if 'rsi' in str(c).lower()])
print("growth_future 관련:", [c for c in df.columns if 'growth_future' in str(c)])
```

```python
# =============================================================================
# Q4 STEP 2~3: RSI < 30 (oversold) 신호를 2000-01-01 ~ 2025-06-01 구간에서 추출
#
# RSI(Relative Strength Index): 0~100 범위로 정규화된 모멘텀 지표
#   RSI < 30 = 과매도 → "너무 떨어졌으니 반등할 것"이라는 평균회귀 베팅
#   RSI > 70 = 과매수
#   timeperiod=14 가 업계 표준
# =============================================================================

# 날짜 타입 보장 (parquet 이 이미 datetime 으로 저장했을 가능성이 높지만 방어)
df['Date'] = pd.to_datetime(df['Date'])

RSI_THRESHOLD = 30
START_DATE = '2000-01-01'
END_DATE   = '2025-06-01'

# pandas 에서 복수 조건은 and/or 가 아니라 &/| + 각 조건 괄호
selected_df = df[
    (df['rsi'] < RSI_THRESHOLD) &
    (df['Date'] >= START_DATE) &
    (df['Date'] <  END_DATE)          # 숙제 표현상 6/1 미만으로 해석
].copy()

print(f"신호 발생 건수: {len(selected_df):,}")        # ← 5,206 근처 기대
print(f"고유 종목 수: {selected_df['Ticker'].nunique()}")
print(f"실제 기간: {selected_df['Date'].min().date()} ~ {selected_df['Date'].max().date()}")
```

```python
# =============================================================================
# Q4 STEP 4: 신호마다 $1,000 투자, 30일 후 청산했을 때의 순이익
#
# 공식: net_income = 1000 * (growth_future_30d - 1).sum()
#   growth_future_30d 는 '비율'(1.05 = +5%)이므로
#   -1 을 해야 순수익률(0.05)이 됨. 안 빼면 원금까지 합산되는 오류
# =============================================================================

GROWTH_COL = 'growth_future_30d'
INVESTMENT = 1000

# ---- 검증값 3개 먼저 확인 ----
n_trades = len(selected_df)
avg_return = selected_df[GROWTH_COL].mean()
win_rate = (selected_df[GROWTH_COL] > 1).mean()

print("숙제가 제시한 검증값과 대조:")
print(f"  거래 건수 : {n_trades:,}        (기대 ≈ 5,206)")
print(f"  평균 수익률: {(avg_return - 1) * 100:.2f}%   (기대 ≈ +1.26%)")
print(f"  승률      : {win_rate * 100:.2f}%   (기대 ≈ 55.13%)\n")

# NaN 확인: 각 종목의 마지막 30일은 미래 데이터가 없어 NaN
n_nan = selected_df[GROWTH_COL].isna().sum()
print(f"growth_future_30d 결측: {n_nan}건 (정산 불가 → .sum() 이 자동 제외)\n")

# ---- 순이익 계산 ----
# .sum() 은 NaN 을 무시하므로 별도 dropna 불필요
net_income = INVESTMENT * (selected_df[GROWTH_COL] - 1).sum()

print(f"▶ 정답 — 순이익: ${net_income:,.0f}")
print(f"          = ${net_income / 1000:.1f} thousands")
print(f"   (총 투입 자본: ${n_trades * INVESTMENT:,} → 수익률 {net_income / (n_trades * INVESTMENT) * 100:.2f}%)")
```

```python
# =============================================================================
# 숙제 Observations 확인 + 전략의 구조적 약점 점검
# =============================================================================

# ---- 1) 임계값 25 → 30 으로 올렸을 때 기회 증가 ----
for thr in [25, 30]:
    sub = df[(df['rsi'] < thr) & (df['Date'] >= START_DATE) & (df['Date'] < END_DATE)]
    ni = INVESTMENT * (sub[GROWTH_COL] - 1).sum()
    print(f"RSI < {thr}: {len(sub):,}건, 순이익 ${ni/1000:.1f}k, "
          f"평균 {(sub[GROWTH_COL].mean()-1)*100:+.2f}%")
print()

# ---- 2) 수익의 분포: 소수 대박이 전체를 캐리하는가? ----
profits = (selected_df[GROWTH_COL] - 1) * INVESTMENT
print("거래별 손익 분포:")
print(profits.describe().round(2).to_string(), '\n')

top10_share = profits.nlargest(10).sum() / profits.sum()
print(f"상위 10건이 전체 이익에서 차지하는 비중: {top10_share*100:.1f}%")
print("→ 비중이 높다면 소수 이상치에 의존하는 취약한 전략\n")

# ---- 3) 포지션 중복 문제 ----
# RSI<30 이 연속으로 뜨면 같은 하락 구간에 중복 베팅하게 됨
dup = selected_df.groupby('Ticker').size().sort_values(ascending=False)
print("종목당 신호 발생 횟수 (상위 10):")
print(dup.head(10).to_string())
print(f"\n→ 한 종목에서 신호가 수십 번 나오면 분산이 아니라 집중 베팅에 가까움")
```
